```python
from bertopic import BERTopic
from sklearn.datasets import fetch_20newsgroups

docs = fetch_20newsgroups(subset='all',  remove=('headers', 'footers', 'quotes'))['data']
```

```python
# Let's see what the docs look like. This was not in the code but when reproducing the example, it is essential to understand the data.
print(len(docs))
print(type(docs))
print(docs[0])
print(type(docs[0]))
print(len(docs[0]))
```

Great, our data is a *list of strings*. It is a pretty big list with 18846 entries.

```python
# Let's extract topics from the documents

topic_model = BERTopic()
topics, probs = topic_model.fit_transform(docs)
```

What did we do here?
- First we initiated an empty `topic_model` of BERTopic class. 
- Then, we called the `fit_transform()` method and pass our `docs` list as the text corpus that the model should be trained on
